```python
# --- Load data ---
csv_path = r'C:\\Users\\bennt\\Desktop\\results_1_32.csv'  # change path as needed
df = pd.read_csv(csv_path)

# --- Extract data ---
y = df.iloc[:, -3]
x = df.iloc[:, -2]
color = df.iloc[:, -1] * 10e-9 * 10e8 * -1  # Convert current scaling

nx = 231  # len(np.unique(x))
ny = 151  # len(np.unique(y))

current_2d = np.array(color).reshape((ny, nx))

# --- Reflect color data about x and y axes ---
# Flip horizontally (mirror across y-axis)
current_2d = np.fliplr(current_2d)
# Flip vertically (mirror across x-axis)
current_2d = np.flipud(current_2d)

extract_bias_point_final(np.array(x), np.array(y), current_2d, minAngleDeg = -50, maxAngleDeg = -40, minLineLength = 10, debug = True)
```
